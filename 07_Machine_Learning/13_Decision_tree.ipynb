```python
import math
```

```python
# we have a dataset elements it has 10 elements
# 6 elements are divided into a variable
# 4 elements are given to b variable
n_A = 6
n_B = 4
total = n_A + n_B
print(total)
```

```python
# Now we find the proportions of elments
prop_A = n_A / total
prop_B = n_B / total
print(prop_A,prop_B)
```

```python
# Now we will find the entropy
# Entropy is a measurement of uncertainity

Entropy = -prop_A * math.log2(prop_A) - prop_B * math.log2(prop_B)
print(Entropy) 
```

# Gini Impurity

```python
# Gini impurity is a measure of misclassification

gini = 1- prop_A**2 - prop_B**2
print(gini)
```

# Information Gain


```python
# In information gain we will assume that a dataset is subdivided into two datasets
# subset of 1: 2 elements of A, 3 elements of B
# subset of 2: 2 elements of A, 3 elements of B
# Entropy and size for each subset
n_1_A, n_1_B = 2,3
n_2_A, n_2_B = 2,3

# find proportions
p_1_A = n_1_A /(n_1_A + n_1_B)
p_1_B = n_1_B /(n_1_A + n_1_B)

# find entropy
entropy_1 = -p_1_A * math.log2(p_1_A) - p_1_B * math.log2(p_1_B) if p_1_A and p_1_B else 0

# find proportions of 2nd subset
p_2_A = n_2_A / (n_2_A + n_2_B)
p_2_B = n_2_B / (n_2_A + n_2_B)

# find entropy
entropy_2 = -p_2_A * math.log2(p_2_A) - p_2_B * math.log2(p_2_B) 

# formula of information gain 
info_gain = Entropy -((n_1_A + n_1_B)/ total * entropy_1 + (n_2_A + n_2_B)/ total * entropy_2)
print(info_gain)
```
