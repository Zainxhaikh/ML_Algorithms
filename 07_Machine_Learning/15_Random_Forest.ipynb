# Random Forest Classifier

### Random forest classifier is a type of supervised machine learning algorithm. You can clearly see it from its name it creates a forest and makes it somehow random.The forest it builds is ensamble most of the time it is trained with bagging method.The general idea of the bagging method is that a combination of learning models increases the over all result.

### YOu can say that Random forest builds multiple decision trees and merges them together to get a mor accurate and stable prediction.

```python
# Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Ml libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import classification_report,confusion_matrix
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor 
```

```python
# Import the dataset
df = sns.load_dataset("tips")
df.head()
```

```python
df.isnull().sum()
df.info()
```

```python
le = LabelEncoder()
for col in df.columns:
    if df[col].dtype =="category":
     df[col] =le.fit_transform(df[col])
df.head()
```

```python
scaler = StandardScaler()
scaler.fit_transform(df)
```

```python
# Split the data into X and y
X = df.drop("tip", axis =1)
y = df["tip"]

# split the X and y for model training
X_train,X_test,y_test,y_train = train_test_split(X,y, test_size=0.2, )

# call and fit the model and predict the model
model = RandomForestRegressor()
model.fit(X_train,y_train)
y_predict = model.predict(X_test)

# Evaluation of the model
print("classification_report :", classification_report(y_test,y_predict))
print("confusion_matrix :", confusion_matrix(y_test,y_predict))
```
