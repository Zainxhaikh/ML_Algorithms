# Linear Regression

### Linear regression is process when we give one variable and it gives output which is continous and and have a constant slop. it has many types. but we discussed here two types.

### 1. simple linear regression
### 2. Multi linear regression

```python
 # import libraries
 
 import pandas as pd
 import numpy as np
 import seaborn as sns
 import matplotlib.pyplot as plt
 from sklearn.preprocessing import StandardScaler
 from sklearn.metrics import mean_squared_error, r2_score
 from sklearn.model_selection import train_test_split
 from sklearn.linear_model import LinearRegression
```

```python
# Load the dataset
df = sns.load_dataset("tips")
df.head()
```

```python
# Scale the data
# scaler = StandardScaler()
# df[["total_bill"]] = scaler.fit_transform(df[["total_bill"]])
```

```python
# split the data
X = df[["total_bill"]]
y = df[["tip"]]
```

```python
# split the data for model training
X_test, X_train, y_test,y_train = train_test_split(X,y, train_size = 0.8)
```

```python
# Call the model
model = LinearRegression()
```

```python
# Fit the model
model = model.fit(X_train,y_train)
```

```python
# slope and intercept check and nmke equation
print("Slope:", model.intercept_)
print("Coefficient:", model.coef_)
print("y=", model.intercept_,"+",model.coef_,"X")
```

```python
# Predict the model
y_predict = model.predict(X_test)
```

```python
 # evaluate the model
 print("MSE :", mean_squared_error(y_test,y_predict))
 print("R2_score :",r2_score(y_test,y_predict))
 print("RMSE :", np.sqrt(mean_squared_error(y_test,y_predict)))
```

```python
# Plot the data to check the model
plt.scatter(X_test,y_test)
plt.plot(X_test,y_predict, color="red")
```
